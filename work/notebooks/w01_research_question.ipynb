## 2. The question: decision, action, cost of a wrong call

## The Question

**Research Question**

Which webpages should be prioritised for content refresh based on observable page characteristics?

**Decision**

Determine whether a webpage should be placed in the content refresh queue.

**Action**

Editors review and update the highest-priority pages first.

**Unit of Analysis**

An individual webpage.

**Output**

A ranked list of webpages based on their likelihood of needing a content refresh.

**Cost of a Wrong Recommendation**

If a declining page is not recommended, it may continue losing visibility and traffic. If a healthy page is incorrectly prioritised, editorial time and effort are spent on content that may not require updating.

**Why Machine Learning Helps**

Multiple observable signals such as impressions, content age, average position, click-through rate, and update history can interact in complex ways. Machine learning can help identify useful patterns that support prioritisation while still allowing human review before any action is taken.

## 3. Quick look at the data (2-3 real numbers)

## Supporting Observations

The starter dataset contains **30,000 webpages**, providing a substantial number of examples for analysing content performance. The correlation between search volume and impressions is approximately **0.001**, suggesting that search volume alone is not a strong predictor of page visibility. Additionally, approximately **54.2%** of the pages are labelled as declining, indicating that content refresh prioritisation is a meaningful decision to investigate using this dataset.

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. Number of pages
print("Rows in dataset:", len(df))

# 2. Correlation between search volume and impressions
corr = df["search_volume"].corr(df["impressions_90d"])
print("Correlation (search_volume vs impressions_90d):", round(corr, 4))

# 3. Percentage of declining pages
declining = (df["trend_direction"].str.lower() == "down").mean() * 100
print("Declining pages:", round(declining, 1), "%")
```
